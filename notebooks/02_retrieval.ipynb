#### Notebook para probar la búsqueda semántica de medicamentos

```python
#----------impotación de librerías y paquetes
from pathlib import Path
import sys
PROJECT_ROOT = Path.cwd().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0,str(PROJECT_ROOT))
from src.ingestion.loader import build_corpus
from src.retrieval.vector_store import VectorStore

import warnings
warnings.filterwarnings("ignore")
```

```python
#-----------construcción del corpus de medicamentos con una muestra de 25000 registros 
df = build_corpus().reset_index(drop=True)
```

```python
#-----------construcción de la metadata de cada medicmaneto
store = VectorStore()
ids = [str(i) for i in df.index]
docs = df["document"].tolist()
metas = [{"name": n, "therapeutic_class":tc}
         for n, tc in zip(df["name"], df["therapeutic_class"])]
n = store.index(ids,docs,metas)
print("------Medicamentos indexados:",n)
```

```python
#-------------prueba de la búsqueda semántica, recuperamos los tres medicamentos más cercanos
preguntas = [
    "medicamento para la presión alta",
    "algo para la alergia",
    "tratamiento para la acidez estomacal"
]

for pregunta in preguntas:
    print(f"---Pregunta: {pregunta}")
    for r in store.search(pregunta,top_k=3):
        print(f"   → {r['metadata']['name']}  (distancia: {r['distance']:.3f})")
        print(r["document"])
    print()
```
